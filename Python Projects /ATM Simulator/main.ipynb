```python
print("WELCOME TO PYBANK ATM")

# account setup
name= input("Enter your username: ").lower()

#PIN setup with validation
while True:
    pin= (input("Set your 4 digit pin: "))
    if pin.isdigit() and len(pin)== 4 :
      print("PIN set successfully.")
      break
    else:
      print("Invalid PIN. PIN must be exactly 4 digits.")

print("---------------------------------------------")

#login section
print("LOGIN (Maximum 3 attempts)")
attempts= 3
login_success= False

#loop for login attempts
while attempts>0:
   user_name= input("Enter Your Username: ").lower()
   user_pin= input("Enter Your 4 digit pin: ")
   if name==user_name and user_pin== pin:
      print("Login Successful!")
      login_success=True
      break
   else:
      attempts-=1
      print(f"Incorrect Details! {attempts} left. ")

#if login fails,stop further execution
if not login_success:
    print("Too many Failed attempts. You are Exiting...")


# receipt function
def print_receipt(name,transaction,amount,balance):
  print("-----RECEIPT-----")
  print("Name        :",name)
  print("Transaction :",transaction)
  print("Amount      :",amount)
  print("Balance     :",balance)

# balance function
balance= 0
def check_balance(balance):
    print(f"Current Balance: ₹{balance:.2f}")

# deposit function
def deposit(balance):
  while True:
    amount= float(input("Enter amount: "))

    #validate deposit amount
    if amount<=0:
      print("Invalid Amount. Enter a Positive Value.")
    else:
      balance += amount
      print(f"₹{amount} deposited Successfully.")
      print(f"Current balance: ₹{balance:.2f}")

      # ask user if they want a receipt
      while True:
        bill_choice=input("Do You want a receipt? (yes/no): ").lower()
        if bill_choice in ['yes','no']:
          break
        print("Invalid Input")

      #print receipt if use chooses yes
      if bill_choice=='yes':
        print_receipt(name,"Deposit",amount,balance)

      return balance

# withdraw function
def withdraw(balance):
  while True:
    amount= float(input("Enter amount: "))

    #validate withdrawal amount
    if amount > balance:
      print("Insufficient Balance.")
    elif amount <=0:
      print("Invalid Amount. Enter a Positive Value.")
    else:
      balance -= amount
      print(f"₹{amount} has Withdrawn Successfully.")
      print(f"Remaining balance: ₹{balance:.2f}")

      # ask user if they want a receipt
      while True:
        bill_choice=input("Do You want a receipt? (yes/no): ").lower()
        if bill_choice in ['yes','no']:
          break
        print("Invalid Input")

      #print receipt if use chooses yes
      if bill_choice=='yes':
        print_receipt(name,"Withdrawal",amount,balance)
      return balance

if login_success:

    # main menu loop until user chooses to exit.
    while True:
      print("---------------------------------------------")
      print("\n MAIN MENU")
      print("1. Check Balance")
      print("2. Deposit Money")
      print("3. Withdraw Money")
      print("4. Exit")

      choice= int(input("Enter Your choice: "))

      #perform operation based on user choice.
      if choice==1:
        check_balance(balance)

      elif choice==2:
        print("DEPOSIT")
        balance= deposit(balance)

      elif choice==3:
        print("WITHDRAW")
        balance= withdraw(balance)

      elif choice==4:
        print(f"Thankyou {name}!")
        print("Transaction Succesfully Completed.")
        break

      else:
        print("Invalid choice. Please Enter a Valid choice Between 1-4.")

```
